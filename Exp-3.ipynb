```python
import pandas as pd
# Load dataset
df = pd.read_csv("data/appointsments_data.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
# Convert datetime
df["appointment_date"] = pd.to_datetime(
    df["appointment_date"],
    format="%d-%m-%Y %H:%M"
)
```

```python
# Convert datetime
df["scheduling_date"] = pd.to_datetime(
    df["scheduling_date"],
    format="%d-%m-%Y"
)
```

```python
# Extract datetime information
df["Date"] = df["appointment_date"].dt.date
df["Hour"] = df["appointment_date"].dt.hour
df["Weekday"] = df["appointment_date"].dt.day_name()
df["Month"] = df["appointment_date"].dt.month_name()
df.head()
```

```python
# Set datetime index
df = df.set_index("appointment_date")
```

```python
# Sort
df = df.sort_index()
```

```python
# Appointments on 5 Nov
print(df.loc["2024-11-05"])
```

```python
print("\nAppointments Every Hour:")
print(df.groupby(df.index.hour).size())
```

```python
# 8. Number of appointments for every weekday
print("\nAppointments Every Weekday:")
print(df.groupby("Weekday").size())
```

```python
# 9. Age Group having the highest appointments
print("\nAppointments by Age Group:")
print(df.groupby("age_group").size().sort_values(ascending=False))
```

```python
# 10. First and last appointment date
print("\nFirst Appointment :", df.index.min())
print("Last Appointment  :", df.index.max())
```
