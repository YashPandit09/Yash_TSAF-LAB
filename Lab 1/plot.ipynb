```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cyclic = pd.read_csv("data/cyclic_component.csv")
cyclic.head()
```

```python
cyclic['date'] = pd.to_datetime(cyclic['date'])
plt.figure(figsize=(12, 6))
plt.plot(cyclic['date'], cyclic['economic_activity'], linewidth=2)
plt.xlabel('Date')
plt.ylabel('Economic Activity')
plt.title('Economic Activity Over Time')



plt.show()
```

Repeating cycles of rise and fall


```python
noise = pd.read_csv("data/irregular_noise_component.csv")
noise['date'] = pd.to_datetime(noise['date'])
noise.head()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(noise['date'], noise['value'])
plt.xlabel('Date')
plt.ylabel('Noise Value')
plt.title('Irregular Noise Component Over Time')
plt.show()
```
