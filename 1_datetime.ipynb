```python
from datetime import datetime
from datetime import date
from datetime import time
from datetime import timedelta
```

```python
#Create Todays date
today = date.today()
print(today)
```

```python
#Create custom date
d = date(2026, 8, 15)
print(d)
```

```python
#Access individual components
print(d.year)
print(d.month)
print(d.day)
```

```python
#Current date and time
from datetime import datetime
now = datetime.now()
print(now)
```

```python
print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)
```

```python
#Formatting dates
print(now.strftime("%d - %m - %Y"))
print(now.strftime("%d / %m / %y"))
print(now.strftime("%d / %m / %Y %H : %M : %S"))
print(now.strftime("%d  %B  %Y"))
print(now.strftime("%A"))
print(now.strftime("%d / %m / %Y %H : %M : %S %p"))
print(now.strftime("%j"))
```

```python
#Convert string to datetime
dob = "15-08-2000"

date_obj = datetime.strptime(dob, "%d-%m-%Y")
print(date_obj)
```

```python
#Date Arithmetic
from datetime import timedelta
today = date.today()
after10days = today + timedelta(days = 10)
print(after10days)
```

```python
joining = date(2026, 7, 14)
today = date.today()
difference = today - joining
print(difference.days)
```

```python
today = date.today()
difference = 3000
doj = today - timedelta(days = difference)
print(doj)
```

```python
#Day of the week
today.strftime("%A")
today.weekday()
```

```python
#Create a Time object
from datetime import time

lecture_time = time(10, 30)

print("Lecture Time: ", lecture_time)
```

```python
#Access Hour, Minute and Second
from datetime import time

meeting = time(14, 45, 30)

print("Hour: ", meeting.hour)
print("Minute: ", meeting.minute)
print("Second: ", meeting.second)
```

```python
#Compare two times
from datetime import time+

office_start = time(9, 0)
office_end = time(17, 30)

print(office_start < office_end)
```

```python
from datetime import datetime, time, date

start = time(9, 30)
end = time(17, 45)

start_dt = datetime.combine(date.today(), start)
end_dt = datetime.combine(date.today(), end)

difference = end_dt - start_dt

print("Difference: ", difference)
print("Hours: ", difference.total_seconds() / 3600)
```
